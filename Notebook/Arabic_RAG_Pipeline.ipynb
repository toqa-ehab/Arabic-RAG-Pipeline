```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "mistralai/Mistral-Nemo-Instruct-2407"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.float16, device_map="auto")
```

```python
def generate_text(prompt, max_length=100, num_return_sequences=1):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(
        **inputs,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        do_sample=True,
        top_k=50,
        top_p=0.95,
        temperature=0.7,
    )
    return [tokenizer.decode(output[inputs["input_ids"].shape[1]:], skip_special_tokens=True) for output in outputs]
```

```python
from sentence_transformers import SentenceTransformer
from PyPDF2 import PdfReader
import faiss

embedding_model = SentenceTransformer(
    "Omartificial-Intelligence-Space/GATE-AraBert-v0"
)
```

```python
def text_from_pdf (path):
    reader=PdfReader(path)
    full_text=""
    for page in reader.pages:
        full_text+=page.extract_text()+ "\n"
    return full_text
```

```python
def chunk_text(text, chunk_size=500, overlap=50):
    words = text.split()
    chunks = []
    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i + chunk_size])
        chunks.append(chunk)
    return chunks
```

```python
def embed_chunks(chunks, model_name="Omartificial-Intelligence-Space/GATE-AraBert-v0"):

    embedding_model = SentenceTransformer(model_name)

    embeddings = embedding_model.encode(
        chunks,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    return embedding_model, embeddings

```

```python
def faiss_index(embeddings):

    dim = embeddings.shape[1]

    index = faiss.IndexFlatIP(dim)

    index.add(embeddings)

    return index
```

```python
def search_index(query, embedding_model, index, chunks, k=5):

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores, indices = index.search(query_embedding, k)

    return [chunks[i] for i in indices[0]]
```

```python
path = "/kaggle/input/datasets/toqaehab/taskkk/ML.pdf"  

full_text = text_from_pdf(path)
chunks = chunk_text(full_text,chunk_size=50, overlap=5)

embedding_model, embeddings = embed_chunks(chunks)

index = faiss_index(embeddings)
```

```python
questions = [
"ما هو التعلم الآلي؟" , 

"ما الفرق بين التعلم الخاضع للإشراف وغير الخاضع للإشراف؟" ,

 "ما هي Embeddings؟"

]
```

```python
for question in questions:

    top_chunks = search_index(
        question,
        embedding_model,
        index,
        chunks,
        k=3
    )

    context = "\n\n".join(top_chunks)

    prompt = f"""
أجب عن السؤال التالي باستخدام المعلومات الموجودة في النص.

السؤال:
{question}

النص:
{context}

الإجابة:
"""

    answer = generate_text(
        prompt,
        max_length=700
    )

    print(f"\nQuestion: {question}")
    print(f"Answer: {answer[0]}")
    print("-" * 50)
```

```python
from fastapi import FastAPI
from pydantic import BaseModel
from fastapi.middleware.cors import CORSMiddleware

app = FastAPI()

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"],
)

class QuestionRequest(BaseModel):
    question: str

@app.post("/ask")
def ask_question(request: QuestionRequest):
    top_chunks = search_index(request.question, embedding_model, index, chunks, k=3)
    context = "\n\n".join(top_chunks)

    prompt = f"""
أجب عن السؤال التالي باستخدام المعلومات الموجودة في النص.

السؤال:
{request.question}

النص:
{context}

الإجابة:
"""
    answer = generate_text(prompt, max_length=700)
    return {"answer": answer[0], "chunks": top_chunks}
```
